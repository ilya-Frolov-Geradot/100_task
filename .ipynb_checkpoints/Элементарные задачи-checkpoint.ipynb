## Чётное или нечётное

```python
num_for_check_1 = 3
num_for_check_2 = 4
if num_for_check_1%2 == 0:
    print('Чётное')
else:
    print('Нечётное')
if num_for_check_2%2 == 0:
    print('Чётное')
else:
    print('Нечётное')
```

```python
num_for_check = int(input('Введите число для проверки: '))
if num_for_check%2 == 0:
    print('Чётное')
else:
    print('Нечётное')
```

## Факториал

```python
f = 3
num_f = 1
for i in range(1, f+1):
    num_f = num_f * i
    i += 1
print(num_f)
```

```python
f = int(input('Введите число для вычесления его факториала: '))
num_f = 1
for i in range(1, f+1):
    num_f = num_f * i
    i += 1
print(num_f)
```
